```python
## creating tuples
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
numbers=tuple([1,2,3,4,5,6])
print(numbers)
```

```python
mixed_tuple=(1,"Hello World",3.14,True)
print(mixed_tuple)
```

```python
## Accessing tuple elements
print(numbers[2])
print(numbers[-1])
print(numbers[0:4])
print(numbers[2:])
print(numbers[::-1])
```

```python
## tuples operation 
concatenation_tuple=numbers + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple *3
```

```python
numbers*3
```

```python
## tuples methods
print(numbers.count(1))
print(numbers.index(4))
```

```python
## packing and unpacking tuple
# packing
packed_tuple=1,"Hello",3.14
print(packed_tuple)
```

```python
# unpacking
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpackinf with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuple
nested_tuple=((1,2,3),("a","b","c"),(True,False))
# accesing the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
  for item in sub_tuple:
    print(item,end=" ")
  print()
```
