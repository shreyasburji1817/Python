```python
## Sets

my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set=set([1,2,3,4,5])
print(my_set)
```

```python
my_set=set([1,2,2,3,4,4,5])
print(my_set)
```

```python
## Basic Set Operations
## adding elements to a set

my_set.add(7)
print(my_set)
```

```python
## removing elements from a set
my_set.remove(4)
print(my_set)
```

```python
my_set.discard(10)
print(my_set)
```

```python
## pop method
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear the set
my_set.clear()
print(my_set)
```

```python
## Set menbership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(6 in my_set)
```

```python
## Mathematical set operations
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
# union
print(set1.union(set2))
# intersection
print(set1.intersection(set2))
# difference
print(set1.difference(set2))
# symmetric difference
print(set1.symmetric_difference(set2))
```

```python
## Sets methods
set1={1,2,3,4,5}
set2={3,4,5}
# is subset
print(set1.issubset(set2))
# is superset
print(set1.issuperset(set2))
```

```python
lst=[1,2,2,3,3,4,5]
set(lst)
```

```python
## Counting unique words in text
text="My name is Shreyas Burji"
words=text.split()
# covert list of words to set to grt unique words
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
