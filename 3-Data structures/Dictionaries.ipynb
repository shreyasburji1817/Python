```python
## creating a Dictionary
empty_dict={}
print(empty_dict)
print(type(empty_dict))
```

```python
empty_dict=dict()
print(empty_dict)
```

```python
student={"name":"Shreyas","Age":23,"grade":24}
print(student)
print(type(student))
```

```python
## accessing dictionary elements
student={"name":"Shreyas","Age":23,"grade":'A'}
print(student)
print(student['name'])
print(student['grade'])
```

```python
## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name','not found'))
```

```python
## Modifying Dictionary elements
student['Age']=33      ## update the value of age key
print(student)
student['address']='INDIA'    ## Add a new key and value
print(student)
del student['age']    ## delete the key and value pair
print(student)
```

```python
## dictionary methods
keys=student.keys()   # get all the keys in the dictionary
print(keys)
values=student.values()  # get all the values in the dictionary
print(values)
items=student.items()  # get all the key-value pairs in the dictionary
print(items)
```

```python
## Shallow copy
student_copy=student
print(student_copy)
print(student)
```

```python
student['name']='Shreyas1'
print(student_copy)
print(student)
```

```python
student_copy1=student.copy()  ## deep copy
print(student_copy1)
print(student)
```

```python
student['name']='Shreyas B'
print(student_copy1)
print(student)
```

```python
## Iterating over Dictionary
# iterating over keys
for keys in student.keys():
  print(keys)
```

```python
# iterating over values
for values in student.values():
  print(values)
  
```

```python
# iterating over key value pairs
for items in student.items():
  print(items)
```

```python
## Nested Dictionary
students={
  "student1":{"name":"Shreyas","age":23},
  "student2":{"name":"Krish","age":32},
}
print(students)
```

```python
## access nested dictionary elements
print(students["student1"]["name"])
print(students["student2"]["age"])
```

```python
## iterating over nested dictionary
for students_id,students_info in students.items():
  print(f"{students_id}:{students_info}")
  for key,vlaue in students_info.items():
    print(f"{key}:{vlaue}")
```

```python
## Dictionary Comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionary comprehension
even={x:x**2 for x in range(10) if x%2==0}
print(even)
```
