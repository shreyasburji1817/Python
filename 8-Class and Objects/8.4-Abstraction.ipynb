### Abstraction

It is a concept of hiding the complex implementation details and showing only necessary features of an object.This help in reducing programming complexity and efforts.

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
  def drive(self):
    print("the vehicle is used for driving")
    
  @abstractmethod
  def start_engine(self):
    pass

class Car(Vehicle):
  def start_engine(self):
    print("the car engine is started")

def operate_vehicle(vehicle: Vehicle):
  vehicle.drive()
  vehicle.start_engine()
car = Car()
operate_vehicle(car)
```
