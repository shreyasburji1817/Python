### Operator Overloading

Operator overloading allows you to define the behavior of operators(+,-,*,etc.) for custom objects.You achieve this by overriding specific magic methods in your class.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __truediv__(self, scalar):
        return Vector(self.x / scalar, self.y / scalar)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)
print(v1 + v2)  # Output: Vector(6, 8)
print(v1 - v2)  # Output: Vector(-2, -2)
print(v1 * 2)   # Output: Vector(4, 6)
print(v1 / 2)   # Output: Vector(1.0, 1.5)
print(v1)       # Output: Vector(2, 3)
print(v2)       # Output: Vector(4, 5)
```
