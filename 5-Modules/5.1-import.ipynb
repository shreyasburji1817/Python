## Importing Modules in Python : Modules and Packages

In python, modules and packages help organize and reuse code.Here's comprehensive guide on how to import them.

```python
import math  # in built module for mathematical operations
math.sqrt(16)  # returns the square root of 16
```

```python
from math import sqrt,pi # import specific functions from the math module
print(sqrt(25))  # returns the square root of 25
print(pi)  # returns the value of pi
```

```python
import sys
print(sys.executable)
```

```python
import numpy as np  # import numpy module for numerical operations
np.array([1, 2, 3])  # creates a numpy array
```

```python
from math import * # import all functions from the math module
print(sqrt(25))  # returns the square root of 25
print(pi)
```

```python
from package.maths import addition  # import the addition function from the maths module
addition(5, 3)  # call the addition function
```

```python
from package import maths
maths.addition(5, 3)  # call the addition function from the maths module
```
