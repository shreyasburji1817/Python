### Standard Library Overview

Python's Standard Library is a vast collection of modules and packages that come bundled with python,provding a wide range of 
functionalities out of the box.Here's an overview of some of the most commonly used modules and packages in the python Standard libraray.

```python
import array 
arr=array.array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
## Random
import random
print(random.randint(1, 10)) # Random integer between 1 and 10
print(random.choice(['apple', 'banana', 'mango'])) # Random choice from a list
```

```python
## File and Directory access
import os 
print(os.getcwd()) # Get current working directory
```

```python
print(os.listdir()) # List files and directories in the current directory
```

```python
print(os.mkdir('new_directory')) # Create a new directory
```

```python
## High level operations on files and collection of files
import shutil  # this module provides a higher level interface for file operations
shutil.copy('source.txt', 'destination.txt')  # Copy a file
```

```python
## data serialization
import json #this module provides functions to serialize and deserialize data in JSON format
data={'name':'Shreyas','age':23}
json_str=json.dumps(data) # Serialize data to JSON string
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str) # Deserialize JSON string back to Python object
print(parsed_data)
print(type(parsed_data))
```

```python
## csv
import csv #this module provides functions to read and write CSV files
with open('data.csv', mode='w', newline='') as file:
  writer=csv.writer(file)
  writer.writerow(['name','age'])
  writer.writerow(['Shreyas',23])

with open('data.csv', mode='r') as file:
  reader=csv.reader(file)
  for row in reader:
    print(row)
```

```python
##datetime 
from datetime import datetime,timedelta #this module provides classes for manipulating dates and times
now=datetime.now() # Get current date and time
print(now)
yesterday=now-timedelta(days=1) # Get yesterday's date and time
print(yesterday)
```

```python
## time
import time #this module provides functions for working with time
print(time.time()) # Get current time in seconds since the epoch
time.sleep(2) # Sleep for 2 seconds
print(time.time()) # Get current time in seconds since the epoch after sleeping for 2 seconds
```

```python
## regular expressions
import re #this module provides functions for working with regular expressions
pattern=r'\d+' # Regular expression pattern to match one or more digits
text='There are 123 apples and 456 oranges.'
match=re.findall(pattern,text) # Find all matches of the pattern in the text
match1=re.search(pattern,text) # Search for the first match of the pattern in the text
print(match) # Print the matched string
print(match1.group()) # Print the matched string
```
