### Standard Library Overview

Python's Standard Library is a vast collection of modules and packages that come bundled with python,provding a wide range of 
functionalities out of the box.Here's an overview of some of the most commonly used modules and packages in the python Standard libraray.

```python
import array 
arr=array.array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
## Random
import random
print(random.randint(1, 10)) # Random integer between 1 and 10
print(random.choice(['apple', 'banana', 'mango'])) # Random choice from a list
```

```python
## File and Directory access
import os 
print(os.getcwd()) # Get current working directory
```

```python
print(os.listdir()) # List files and directories in the current directory
```

```python
print(os.mkdir('new_directory')) # Create a new directory
```

```python
## High level operations on files and collection of files
import shutil  # this module provides a higher level interface for file operations
shutil.copy('source.txt', 'destination.txt')  # Copy a file
```
