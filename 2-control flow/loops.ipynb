```python
## for loop
for i in range(5):
  print(i)
```

```python
for i in range(1,5):
  print(i)
```

```python
for i in range(1,15,3):
  print(i)
```

```python
for i in range(15,1,-2):
  print(i)
```

```python
## while loop
count=0
while count<=5:
  print(count)
  count=count+1
```

```python
## break statement
for i in range(10):
  if i==5:
    break
  print(i)
```

```python
## continue statement
for i in range(10):
  if i%2==0:
    continue
  print(i)
```

```python
## pass statemnt
for i in range(5):
  if i==3:
    pass
  print(i)
```

```python
## nested statement
for i in range(3):
  for j in range(2):
    print(f"i:{i} and j:{j}")
```

```python
## calaculate the sum of first n natural numbers using a while and for loop
## using while loop
n=10
sum=0
count=1
while count<=n:
  sum=sum+count
  count=count+1
print("the sum of first n natural numbers is:",sum)
```

```python
## for loop
n=10
sum=0
for i in range(n+1):
  sum=sum+i
print(sum)
```

```python
## prime numbers between 1 to 100
for num in range(1, 101):
  if num >1:
    for i in range (2, num):
      if num%i==0:
         break
    else:
      print(num)
```
