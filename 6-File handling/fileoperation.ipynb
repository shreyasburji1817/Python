### File Operation- Read and Weite Files

File handling is a 

```python
## read a whole file
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## read a file line by line
with open('example.txt','r') as file:
  for line in file:
    print(line.strip()) # strip() removes the newline character at the end of each line
```

```python
## writing a file (overwriting)
with open('example.txt','w')as file:
  file.write('Hello World\n')
  file.write('This is a test file.\n')
```

```python
## write a file(without overwriting)
with open('example.txt','a')as file:
  file.write('This line will be appended to the file.\n')
  file.write('This is another appended line.\n')
```

```python
### writing a list of lines to a file
lines=['First line\n', 'Second line\n','Third line\n']
with open('example.txt','a')as file:
  file.writelines(lines)
```

```python
### Binary files
## writng to a binary file
data=b'\x00\x01\x02\x03\x04\x05'
with open('example.bin','wb')as file:
  file.write(data)
```

```python
## reading a binary file
with open('example.bin','rb')as file:
    content=file.read()
    print(content)
```

```python
### read the content form a source text file and write to a destination text file
## copy a text file
with open('example.txt','r')as source_file:
  content=source_file.read()

with open('destination.txt','w')as destination_file:
  destination_file.write(content)
```

```python
### read a text file and count the number of lines,words& characters
## counting lines, words and characters in atext file
def content_text_file(file_path):
  with open(file_path,'r')as file:
    lines=file.readlines()
    line_count =len(lines)
    word_count =sum(len(line.split())for line in lines) 
    char_count=sum(len(line)for line in lines)
  return line_count,word_count,char_count

file_path='example.txt'
lines,words,characters=content_text_file(file_path)
print(f'lines:{lines},words:{words},characters:{characters}')
```

The w+ mode in python is used to open a file for both reading and writing.If the file does not exist,
it will be created.If the file exists,its content is trolcated(i.e, the file is overwritten). 

```python
## writing and then reading a file
with open('example.txt','w+')as file:
  file.write("Hello World\n")
  file.write("This is a new line\n")
  ## move the file cursor to the beginnning
  file.seek(0)
  ## read the content of the file
  content=file.read()
  print(content)
```
