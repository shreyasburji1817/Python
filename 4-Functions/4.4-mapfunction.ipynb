The map() Function in Python

The map() function applies a given function to all items in an input list (or any other iterable) and return a map 
object(an iterable).This is particularly useful for transforming data in alist comprehensively.

## Syntax
map(function, iterable)

```python
def squares(num):
  return num*num
squares(5)
```

```python
numbers=[1,2,3,4,5]
list(map(squares,numbers))
```

```python
##lambda function with map
numbers=[1,2,3,4,5]
list(map(lambda num:num**2,numbers))
```

```python
## map multiple iterables
numbers1=[1,2,3]
numbers2=[4,5,6]
added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
## map() to convert a list of strings to integers
numbers=["1","2","3"]
int_numbers=list(map(int,numbers))
print(int_numbers)
```

```python
words=['apple','banana','mango']
upper_words=list(map(str.upper,words))
print(upper_words)
```

```python
def get_names(person):
  return person['name']

people=[
  {'name':'Shreyas','age':23},
  {'name':'Rohit','age':25},
]

list(map(get_names,people))
```
