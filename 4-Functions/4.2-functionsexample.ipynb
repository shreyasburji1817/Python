Example 1: Temperature Converstion 

```python
def convert_tempertaure(temp,unit):
  if unit=='C':
    return temp * 9/5 + 32 ## celsius to fahrenheit
  elif unit=='F':
    return (temp - 32) * 5/9 ## fahrenheit to celsius
  else:
    return None

print(convert_tempertaure(100,'C'))
print(convert_tempertaure(212,'F'))
```

Example 2: Password Strength Checker

```python
def is_strong_password(password):
  if len(password)<8:
    return False
  if not any(char.isdigit() for char in password):
    return False
  if not any(char.isupper() for char in password):
      return False
  if not any(char.islower() for char in password):
      return False
  if not any(char in '!@#$%^&*()_+' for char in password):
      return False
  return True

print(is_strong_password('Password123!'))
```

Example 3: Calculate the Total Cost of Items in a Shopping Cart

```python
cart=[
  {'name':'Apple','price':20,'quantity':4},
  {'name':'Banana','price':15,'quantity':6},
  {'name':'Mango','price':40,'quantity':10},
]

def calculate_total_cost(cart):
  total_cost=0
  for item in cart:
    total_cost+=item['price']*item['quantity']
  return total_cost

print(calculate_total_cost(cart))
```

Example 4: Check IF a string is Palinderome

```python
def is_palindrome(string):
  string=string.lower().replace(' ','')
  return string==string[::-1]
print(is_palindrome('A man a plan a canal Panama'))
print(is_palindrome('Hello World'))
```

Example 5: Calculate the factioral of anumber using recursion

```python
def factorial(n):
  if n==0:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(6))
```
