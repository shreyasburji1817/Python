Lambda Functions in Python

Lambda function are small anonymos functions defined using lamdba keyword.They can have any number of arguments but only 
one expression.They are commonly used for short operations or as arguments to higher-order functions. 

```python
## Syntax
lambda arguments: expression
```

```python
addition=lambda a,b:a+b
print(addition(4,5))
type(addition)
```

```python
even=lambda a: a%2==0
print(even(5))
```

```python
## using map() - applies a function to all the items in an input list
numbers=[1,2,3,4,5]
list(map(lambda x:x**2,numbers))
```
