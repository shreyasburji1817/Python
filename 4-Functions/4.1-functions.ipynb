```python
def even_or_odd(num):
  if num%2==0:
    print("the number is even")
  else:
    print("the number is odd")
```

```python
## default parameters
def greet(name):
  print(f"Hello {name} Welcome to the world of Python")

greet('Shreyas')
```

```python
def greet(name='Guest'):
  print(f"Hello {name} Welcome to the world of Python")

greet()
```

```python
## Variable length arguments
# positional arguments
def print_numbers(* args):
  for numbers in args:
    print(numbers)

```

```python
print_numbers(1,2,3,4,5,6,"Shreyas")
```

```python
# keywords arguments
def print_details(**kwargs):
  for key , value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(name="Shreyas", age=23, country="INDIA")
```

```python
## both in one function
def print_info(*args, **kwargs):
  for val in args:
    print(f"postional arguments:{val}")
  for key , value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_info(1,2,3,4,5,6,"Shreyas", name="Shreyas", age=23, country="INDIA")
```
