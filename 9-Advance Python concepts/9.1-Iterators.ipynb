### Iterators

They are advance python concepts that allow for efficient looping and memory management.Iterators provide a way to access elements of a collection sequentialy without exposing the underlying structure.

```python
my_list=[1,2,3,4,5,6]
for i in my_list:
  print(i)
```

```python
## iterator
iterator=iter(my_list)
print(type(iterator))
```

```python
## iterate through all the elements
next(iterator)
```

```python
my_list=[1,2,3,4,5,6]
iterator=iter(my_list)
try:
  print(next(iterator))
except StopIteration:
  print("End of iterator reached")
```

```python
## string iterator 
my_string="Hello"
string_iterator=iter(my_string)

print(next(string_iterator))  # Output: 'H'
print(next(string_iterator))  # Output: 'e'
```
